```python
#@title { vertical-output: true }
import pandas as pd

path = "data/raw/historic/2025-26/merged_gw.csv"

df = pd.read_csv(path, low_memory=False)

dupes = df[
    df.duplicated(
        ["GW", "fixture", "element"],
        keep=False
    )
].sort_values(
    ["element", "GW", "fixture"]
)

print(f"Duplicate rows: {len(dupes):,}")
print(
    f"Duplicate player-fixtures: "
    f"{dupes[['GW','fixture','element']].drop_duplicates().shape[0]:,}"
)

display(
    dupes[
        dupes["element"].isin(
            dupes["element"].drop_duplicates().head(3)
        )
    ].head(30)
)
```

```python
#@title { vertical-output: true }
# ---------------------------------------------------------
# Are duplicate rows completely identical?
# ---------------------------------------------------------

exact_duplicates = df.duplicated(keep="first")

print(
    f"Exact duplicate rows removable: "
    f"{exact_duplicates.sum():,}"
)

df_exact_deduped = df.drop_duplicates()

remaining = df_exact_deduped[
    df_exact_deduped.duplicated(
        ["GW", "fixture", "element"],
        keep=False
    )
]

print(
    f"Duplicate player-fixtures remaining "
    f"after exact dedupe: {len(remaining):,}"
)
```

```python
#@title { vertical-output: true }
import pandas as pd
import numpy as np

PATH = "data/processed/fct_gw_historic.csv"

df = pd.read_csv(PATH, low_memory=False)

print("=== BASIC SHAPE ===")
print(f"Rows:             {len(df):,}")
print(f"Columns:          {len(df.columns):,}")
print(f"Players:          {df['player_code'].nunique():,}")
print(f"Seasons:          {df['season'].nunique():,}")
print(f"Fixtures:         {df[['season','fixture_id']].drop_duplicates().shape[0]:,}")

display(df.head())
```

```python
#@title { vertical-output: true }
grain = [
    "season",
    "gameweek",
    "player_code",
    "fixture_id"
]

duplicates = df[df.duplicated(grain, keep=False)]

print("=== GRAIN CHECK ===")

if len(duplicates) == 0:
    print("PASS - no duplicate player-fixture rows")
else:
    print(f"FAIL - {len(duplicates):,} duplicate rows")
    display(
        duplicates.sort_values(grain).head(50)
    )
```

```python
#@title { vertical-output: true }
critical_columns = [
    "season",
    "gameweek",
    "fixture_id",
    "player_code",
    "position",
    "team_name",
    "opponent_team_id",
    "minutes",
    "total_points"
]

null_check = (
    df[critical_columns]
    .isna()
    .sum()
    .to_frame("null_rows")
)

null_check["null_pct"] = (
    null_check["null_rows"] / len(df) * 100
).round(2)

display(null_check)
```

```python
#@title { vertical-output: true }
season_summary = (
    df
    .groupby("season")
    .agg(
        rows=("player_code", "size"),
        players=("player_code", "nunique"),
        gameweeks=("gameweek", "nunique"),
        min_gw=("gameweek", "min"),
        max_gw=("gameweek", "max"),
        fixtures=("fixture_id", "nunique")
    )
    .reset_index()
)

display(season_summary)
```

```python
#@title { vertical-output: true }
print("=== TEAM MAPPING ===")

print(
    "Missing team:",
    df["team_name"].isna().sum()
)

print(
    "Missing opponent:",
    df["opponent_team_name"].isna().sum()
)

same_team = df[
    df["team_id"] == df["opponent_team_id"]
]

print(
    "Player team = opponent:",
    len(same_team)
)
```

```python
#@title { vertical-output: true }
position_check = (
    df
    .groupby(["season", "position"])
    .agg(
        rows=("player_code", "size"),
        players=("player_code", "nunique")
    )
    .reset_index()
)

display(position_check)
```

```python
#@title { vertical-output: true }
print("=== MINUTES ===")

display(
    df["minutes"]
    .describe(
        percentiles=[
            .01, .05, .25, .5,
            .75, .95, .99
        ]
    )
)

bad_minutes = df[
    (df["minutes"] < 0) |
    (df["minutes"] > 120)
]

print(
    f"Obviously invalid minutes: {len(bad_minutes):,}"
)
```

```python
#@title { vertical-output: true }
display(
    df["total_points"]
    .describe(
        percentiles=[
            .01, .05, .25, .5,
            .75, .95, .99
        ]
    )
)

print("\nHighest scores:")

display(
    df[
        [
            "season",
            "gameweek",
            "player_name",
            "position",
            "minutes",
            "total_points"
        ]
    ]
    .sort_values(
        "total_points",
        ascending=False
    )
    .head(20)
)

print("\nLowest scores:")

display(
    df[
        [
            "season",
            "gameweek",
            "player_name",
            "position",
            "minutes",
            "total_points"
        ]
    ]
    .sort_values("total_points")
    .head(20)
)
```

```python
#@title { vertical-output: true }
expected_columns = [
    "expected_goals",
    "expected_assists",
    "expected_goal_involvements",
    "expected_goals_conceded"
]

availability = []

for season, season_df in df.groupby("season"):

    row = {
        "season": season,
        "rows": len(season_df)
    }

    for col in expected_columns:

        if col in season_df.columns:
            row[f"{col}_pct"] = round(
                season_df[col].notna().mean() * 100,
                1
            )
        else:
            row[f"{col}_pct"] = 0

    availability.append(row)

availability = pd.DataFrame(availability)

display(availability)
```

```python
#@title { vertical-output: true }
defcon_summary = (
    df
    .groupby("season")
    .agg(
        rows=("player_code", "size"),
        defcon_values=(
            "defensive_contribution",
            lambda x: x.notna().sum()
        ),
        defcon_hits=(
            "defcon_hit",
            lambda x: (x == 1).sum()
        )
    )
    .reset_index()
)

defcon_summary["availability_pct"] = (
    defcon_summary["defcon_values"]
    / defcon_summary["rows"]
    * 100
).round(1)

display(defcon_summary)
```

```python
#@title { vertical-output: true }
player_gw = (
    df
    .groupby(
        ["season", "gameweek", "player_code"]
    )
    .size()
    .reset_index(name="fixtures")
)

dgws = player_gw[
    player_gw["fixtures"] > 1
]

print(
    f"Player-gameweeks with >1 fixture: "
    f"{len(dgws):,}"
)

display(
    dgws
    .sort_values(
        ["season", "gameweek", "fixtures"],
        ascending=[True, True, False]
    )
    .head(50)
)
```

```python
#@title { vertical-output: true }
meaningful = df[
    df["minutes"] >= 45
]

summary_45 = (
    meaningful
    .groupby("season")
    .agg(
        observations=("player_code", "size"),
        players=("player_code", "nunique"),
        avg_minutes=("minutes", "mean"),
        avg_points=("total_points", "mean")
    )
    .reset_index()
)

display(summary_45)
```
